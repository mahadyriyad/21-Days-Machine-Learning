```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

dataset load

```python
dataset= pd.read_csv('/content/dataset.csv')
```

```python
dataset.head()
```

```python
dataset=dataset.drop(['car_ID'], axis='columns')
```

```python
dataset.head()
```

basic inspection

```python
dataset.shape
```

```python
dataset.info()
```

segergate

```python
x_data= dataset.drop(['price'], axis='columns')
x_data.head()
```

```python
x=x_data.select_dtypes(exclude=['object'])
x.head()
```

```python
x.shape
```

```python
y=dataset.iloc[:,-1]
y.head()
```

Scaling the Independent Variables (Features)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = pd.DataFrame(scaler.fit_transform(x), columns=x.columns)
```

```python
X.head()
```

split

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=0)
```

train

# Model-1:RANDOM FOREST REGRESSION

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
model.fit(x_train, y_train)
```

```python
ypred = model.predict(x_test)
```

Evaluating Model

```python
from sklearn.metrics import r2_score
r2score = r2_score(y_test,ypred)
print("R2Score",r2score*100)
```

# Model-2:DECISION TREE REGRESSION

Training Dataset using Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model.fit(x_train, y_train)
```

```python
ypred = model.predict(x_test)
```

evaluation

```python
from sklearn.metrics import r2_score
r2score = r2_score(y_test,ypred)
print("R2Score",r2score*100)
```

# Model-3:POLYNOMIAL REGRESSION

Convert X to Polynomial Format (X^n)


```python
from sklearn.preprocessing import PolynomialFeatures
modelPR = PolynomialFeatures(degree = 3)
xPoly = modelPR.fit_transform(X)
```

Train same Linear Regression with X-Polynomial instead of X

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xPoly, y)
```

```python
y_pred = model.predict(xPoly)
```

Evaluation

```python
r2 = r2_score(y, y_pred)
print("R² Score:", round(r2,4))
```
